```python
using System;
using System.Linq;

public static string[] RenderShape(bool[,] bits, char symbol = '*', char clearSymbol = ' ')
{
    int rows = bits.GetLength(0);
    int cols = bits.GetLength(1);
    string[] result = new string[rows];

    for (int i = 0; i < rows; i++)
    {
        char[] rowChars = new char[cols];
        for (int j = 0; j < cols; j++)
        {
            rowChars[j] = bits[i, j] ? symbol : clearSymbol;
        }
        result[i] = new string(rowChars);
    }
    return result;
}

public static void PrintShape(string[] shape)
{
    foreach (var line in shape)
    {
        Console.WriteLine(line);
    }
}

public static string[] GetRectangle(int width, int height, char symbol = '*', bool isFill = true, char clearSymbol = ' ', int borderWidth = 1)
{
    var bits = new bool[height, width];
    for (int i = 0; i < height; i++)
    {
        for (int j = 0; j < width; j++)
        {
            if (isFill)
            {
                bits[i, j] = true;
            }
            else
            {
                bits[i, j] = (i < borderWidth || i >= height - borderWidth || 
                              j < borderWidth || j >= width - borderWidth);
            }
        }
    }
    return RenderShape(bits, symbol, clearSymbol);
}

public static string[] GetPyramid(int height, char symbol = '*', char clearSymbol = ' ')
{
    int width = height * 2 - 1;
    var bits = new bool[height, width];
    
    for (int i = 0; i < height; i++)
    {
        int start = height - 1 - i;
        int end = height - 1 + i;
        for (int j = start; j <= end; j++)
        {
            bits[i, j] = true;
        }
    }
    return RenderShape(bits, symbol, clearSymbol);
}

public static string[] GetRhombus(int size, char symbol = '*', char clearSymbol = ' ')
{
    int height = size * 2 + 1;
    int width = size * 2 + 1;
    var bits = new bool[height, width];
    
    for (int i = 0; i < height; i++)
    {
        int distFromCenter = Math.Abs(i - size);
        int start = distFromCenter;
        int end = width - 1 - distFromCenter;
        
        for (int j = start; j <= end; j++)
        {
            bits[i, j] = true;
        }
    }
    return RenderShape(bits, symbol, clearSymbol);
}
```
